### 14. Inner vs Outer Joins

```python
pd.merge(ivies, regions, on='School Name')
```

```python
pd.merge(
    ivies, 
    regions, 
    on='School Name',
    how="outer"
).sort_values(by=['School Type'])
```

### 15. Left vs Right Joins

```python
pd.merge(
    ivies, 
    regions, 
    on='School Name',
    how="left"
).sort_values(by=['School Type'])
```

```python
pd.merge(
    ivies, 
    regions, 
    on='School Name',
    how="right"
).sort_values(by=['Starting Median Salary'])
```

### 16. One-to-One and One-to-Many Joins

##### **One to One**

```python
check_one_to_one = regions.loc[:,'School Name']\
    .isin(ivies.loc[:,'School Name'])
regions[check_one_to_one].loc[:,'School Name']\
    .is_unique
```

```python
pd.merge(
    ivies,
    regions,
    on='School Name'
)
```

##### **One to Many**

```python
check_one_to_many = regions.loc[:,'School Name']\
    .isin(state.loc[:,'School Name'])
regions[check_one_to_many].loc[:,'School Name']\
    .is_unique
```

```python
pd.merge(
    state,
    regions,
    on='School Name'
)
```

```python
temp2 = pd.merge(
    state,
    regions,
    on='School Name'
)
```

```python
temp2.loc[:,'School Name'].value_counts()
```

```python
temp2.loc[:,'School Name'].is_unique()
```
