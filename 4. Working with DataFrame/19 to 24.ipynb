## 26. Part II Merging Units with colums Names

```python
units
```

```python
units.info(False)
```

```python
units = units.replace('',np.nan).dropna()
```

```python
units.info()
```

```python
# Drop type(int) colms
units = units.drop(['calories','lucopene'],axis=1)
units
```

```python
units=units.iloc[0,:]
```

```python
units
```

```python
col_with_units= {}
for i in units.index:
    # print(i,": ",units[i])
    col_with_units[i]=i+'_'+units[i]
col_with_units
```

```python
df_nu = df.rename(columns=col_with_units)
```
