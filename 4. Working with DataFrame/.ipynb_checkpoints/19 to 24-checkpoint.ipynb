## 20. The astype() method

```python
df2 = pd.DataFrame({
    'age':[12,13,14,16],
    'weight':[41.1, 34.5, 55.6, 47.3],
    'height':['1.72','1.78','1.52','1.79'],
})
df2
```

```python
df2.info()
```

```python
df2=df2.astype(float)
```

```python
df2.info()
```

```python
df2=df2.astype({'age':int})
```

```python
df2.info()
```
