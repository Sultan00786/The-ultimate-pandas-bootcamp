```python
import numpy as np
import pandas as pd
data_url = 'https://andybek.com/pandas-nutrition'
df = pd.read_csv(data_url)
df=df.drop('Unnamed: 0', axis=1)
df.head()
```

### 1. Find the 10 food that have most Vitamin B12. What do they have in common?

#### Need to convert in float

```python
df.info()
```

```python
without_u_df = df.iloc[:,1:].replace('[a-zA-Z\s]','',regex=True).astype(float)
```

```python
without_u_df.info(False)
```

```python
without_u_df
```

```python
df.iloc[:,[0]]
```

```python
df_nu = pd.concat([df.iloc[:,[0]],without_u_df],axis=1)
```

```python
df_nu.info()
```

#### Now, we can solve the problem

```python
df_nu.sort_values(by=['name','vitamin_b12'],ascending=False).head(10)
```

```python
df_nu.sort_values(by=['vitamin_b12'],ascending=False)\
    .head(10)\
    .filter(items=['name','vitamin_b12'],axis=1)
```
