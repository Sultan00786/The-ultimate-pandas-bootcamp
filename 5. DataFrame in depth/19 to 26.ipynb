### 19. Identifying the Duplicates

```python
import pandas as pd
import numpy as np
data_url = 'https://andybek.com/pandas-soccer'
players = pd.read_csv(data_url)
players.head(6)
```

```python
players[players.duplicated()]
```

```python
# the subset params
```

```python
# unique -> club, age, position, market_value
```

```python
players.duplicated(subset=['club','age','position', 'market_value'])
```

```python
players[players.duplicated(subset=['club','age','position', 'market_value'])]
```

```python
players[players.duplicated(subset=['club','age','position', 'market_value'], keep='last')]
```

```python
players[players.duplicated(subset=['club','age','position', 'market_value'], keep=False)] ## keep=False means --> all dups
```

### 20. Removing the Duplicates

```python
players[players.duplicated()]
```

```python
players.market_value.mean()
```

```python
without_dup.market_value.mean()
```

### 26. Drop And Filling Df NA's

```python
players[players.isnull().values]
```

```python
players.fillna("hellow there").iloc[[30, 192, 195]] ## filling with generic value --> "hellow there"
```
