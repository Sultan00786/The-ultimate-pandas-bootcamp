## 30. Calculating Aggregates With agg()

```python
import pandas as pd
import numpy as np
data_url = 'https://andybek.com/pandas-soccer'
players = pd.read_csv(data_url)
players.head(6)
```

```python
players.info()
```

### extract the dtype **int** and **float**, Method - I

```python
players.age.dtypes == 'int64'
```

```python
players.fpl_value.dtypes == 'float64'
```

```python
players.axes[1]
players['name'].dtype == object
```

```python
num_col = []
for i in players.axes[1]:
    num_col.append(True) if (players[i].dtypes == 'int64') or (players[i].dtypes == 'float64') else num_col.append(False)
num_col
```

```python
players.loc[:,num_col].shape
```

```python
players.loc[:,num_col].agg('sum')
```

### extract the dtype **int** and **float**, Method - II Easy

```python
np.number
```

```python
players.select_dtypes(['int','float64']).agg('sum')
```

```python
players.select_dtypes(np.number).agg('sum')
```

```python
players.select_dtypes(np.number).agg(['sum','max','min'])
```

## 32. More Flexibility wity apply()

```python
players.select_dtypes(['float']).head()
```

```python
def round_off_float(x):
    if x.dtype == 'float64':
        return round(x)
    return x
```

```python
players.select_dtypes(['float']).apply(round_off_float).head()
```

```python
# also can do aggeregated tasks
players.select_dtypes(np.number).apply('mean')
```

```python
# flipping the axis
```

```python
# according to indexs
players.select_dtypes(np.number).apply('mean',axis=1)
```
