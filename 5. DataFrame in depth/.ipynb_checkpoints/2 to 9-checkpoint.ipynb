## 9. Skill Challenge

```python
players.head()
```

```python
players.info()
```

```python
# Find the players in our dataset that meet thees criteria
    # 1. they're English (nationality), and
    # 2. their market value is more that twice the average market value in the league (market_value), andvv
    # 3. they either have more than 4000 views (page_Views) or are a new signing (new_signing) but not both 
```

### 1.

```python
players.nationality.unique()
```

```python
 # 1. they're English (nationality)
English_player = players.nationality == 'England'
players[English_player].filter(items=['name','age','nationality'],axis=1)\
    .head()
```

### 2.

```python
avg_value = players.market_value.mean()
bool_avg_value = players.market_value >= (2*avg_value)
```

```python
 # 2. their market value is more that twice the average market value in the league (market_value)
players[bool_avg_value].filter(items=['name','market_value'], axis = 1)\
    .sort_values(by=['market_value'])\
    .head()
```

### 3.

```python
players.new_signing.unique()
```

```python
exor = (players.new_signing == 1) ^ (players.page_views > 4000)
exor
```

```python
# 3. they either have more than 4000 views (page_Views) or are a new signing (new_signing) but not both 
players[exor].filter(items=['name','new_signing','page_views'], axis=1)
```

### Final solution:

```python
players[English_player & bool_avg_value & exor]\
    .filter(items=['name','nationality','market_value','page_views','new_signing'])
```
