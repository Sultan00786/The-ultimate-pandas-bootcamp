## 3. Quick Review: Indexing with Boolean Mask

```python
players.head()
```

```python
# Q: What are the players that have market value exceeding the 40M?
```

```python
players[players.loc[:,'market_value']>40]
```

## 4. More Approaches to Boolean Masking

```python
players.head()
```

```python
players.loc[:,'position'].unique()
```

```python
# Q: Find all defenders from the player
# Defenders codes: LB, CB, RB
```

```python
df_bool = players.loc[:,'position']\
                .isin(['LB','CB', 'RB'])
df_bool
```

```python
players[df_bool]
```

```python
players.loc[:,'position'].value_counts()
```

```python
# between() method 
```

```python
players.loc[:,'market_value'].between(40,50, inclusive='neither')
```

```python
players[players.loc[:,'market_value'].between(40,50,inclusive='neither')]
```
