### 2. Reading In Data With read_csv()

```python
import pandas as pd
```

```python
pd.read_csv('https://www.andybek.com/pandas-drinks')
```

```python
pd.read_csv('https://www.andybek.com/pandas-drinks', usecols=['country', 'wine_servings'])
```

```python
alcohol = pd.read_csv('https://www.andybek.com/pandas-drinks', usecols=['country', 'wine_servings'], index_col='country')
```
