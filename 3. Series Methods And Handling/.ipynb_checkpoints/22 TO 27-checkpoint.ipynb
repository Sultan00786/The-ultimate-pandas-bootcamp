## 22. Series Arithmetics

```python
import pandas as pd
alcohol = pd.read_csv('https://www.andybek.com/pandas-drinks', usecols=['country', 'wine_servings'], index_col='country').squeeze("columns")
alcohol.head(10)
```

```python
alcohol=alcohol.sort_index()
alcohol.head()
```

```python
increase_drinks = pd.Series({'Albania':6,'Andorra':8, 'Afghanistan':100})
increase_drinks
```

```python
(alcohol+increase_drinks)
```

```python
(alcohol+increase_drinks)[(alcohol+increase_drinks).notnull()]
```

```python
alcohol[(alcohol+increase_drinks).notnull()]
```

```python
alcohol.add(increase_drinks, fill_value=0)
```

## 26. Series Iteration

```python
mini_alco = alcohol.head(10)
mini_alco
```

```python
for i in mini_alco: print(i)
```

```python
for i in mini_alco.index: print(i, " ", mini_alco[i])
```

```python
for i in mini_alco.items(): print(i)
```

## 27. Filtering filter(), where() and mask()

```python
alcohol.filter(regex='^V')
```

```python
alcohol.filter(regex='stan')
```

```python
alcohol.filter(like='stan')
```

```python
alcohol[alcohol<=100]
```

```python
alcohol[alcohol<=100].sort_values()
```

```python
def get100(x):
    return alcohol<=100

alcohol[get100]
```

```python
alcohol.where(lambda x: x<=100)
```

```python
alcohol.where(lambda x: x<=100)
```

```python
alcohol.where(lambda x: x<=100).dropna()
```

```python
alcohol.where(lambda x: ~(x<=100)).dropna()
```
