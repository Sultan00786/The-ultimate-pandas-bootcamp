## 20. Skill Challenge

### 1. select wine from alcohol that have more that 50

```python
import pandas as pd
alcohol = pd.read_csv('https://www.andybek.com/pandas-drinks', usecols=['country', 'wine_servings'], index_col='country').squeeze("columns")
alcohol.head(10)
```

```python
fifty_plus = alcohol[alcohol>50]
fifty_plus.count()
```

### 2. Choose contries which have smallest 20 wine

```python
fifty_plus.sort_values().head(20)
```

```python
fifty_plus.nsmallest(20)
```

### 3. calculate mean, median and standard deviation of step 2

```python
fifty_plus.sort_values().head(20).mean()
```

```python
fifty_plus.sort_values().head(20).median()
```

```python
fifty_plus.sort_values().head(20).std()
```
