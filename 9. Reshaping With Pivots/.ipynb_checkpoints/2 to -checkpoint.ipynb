### 2. New Data

```python
import pandas as pd
import numpy as np

sat_df = pd.read_csv(r"https://andybek.com/pandas-sat")
sat_df.head()
```

```python
sat_df.info()
```

```python
# remove % from Percent Tested column
sat_df.loc[:,'Percent Tested'].replace('%','',regex=True).astype(float)
```

```python
# remove % from Percent Tested column
sat_df.loc[:,'Percent Tested'] = sat_df.loc[:,'Percent Tested'].str.replace('%','',regex=True).astype(float)
sat_df.loc[:,'Percent Tested']
```

```python
sat_df['Percent Tested'].map(type).value_counts()
```

```python
sat_df['Percent Tested'] = pd.to_numeric(
    sat_df['Percent Tested']
)
```

```python
sat_df.info()
```
