### 2. New Data

```python
import pandas as pd
import numpy as np

sat_df = pd.read_csv(r"https://andybek.com/pandas-sat")
sat_df.head()
```

```python
sat_df.info()
```

```python
# remove % from Percent Tested column
sat_df.loc[:,'Percent Tested'].replace('%','',regex=True).astype(float)
```

```python
# remove % from Percent Tested column
sat_df.loc[:,'Percent Tested'] = sat_df.loc[:,'Percent Tested'].str.replace('%','',regex=True).astype(float)
sat_df.loc[:,'Percent Tested']
```

```python
sat_df['Percent Tested'].map(type).value_counts()
```

```python
sat_df['Percent Tested'] = pd.to_numeric(
    sat_df['Percent Tested']
)
```

```python
sat_df.info()
```

### 6. The pivot_table()

```python
sat_df.head()
```

#### What is average Borough score as per Sat section 

```python
sat_df.pivot_table(index='Borough', columns='SAT Section', values='Score') # isko duplicates chalta
```

```python
sat_df.pivot_table(index='Borough', columns='SAT Section', values='Score', aggfunc='mean') # aggrigate fuction
```

```python
sat_df.pivot_table(index='Borough', columns='SAT Section', values='Score', aggfunc=['sum','count']) # aggrigate fuction
```

#### What is average Percent Tested for Borough?

```python
sat_df\
    .pivot_table(index='Borough',values='Percent Tested')\
    .sort_values(by='Percent Tested')
```

### 8. Replicating Pivot Tables With GroupBy

```python
sat_df.head()
```

```python
sat_df.pivot_table(index='Borough', columns='SAT Section', values='Score')
```

```python
sat_df.groupby(['Borough','SAT Section']).agg({'Score':'mean'})
```

```python
sat_df.groupby(['Borough','SAT Section']).agg({'Score':'mean'}).unstack(level=1)
```
