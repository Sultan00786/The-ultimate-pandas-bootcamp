##### Q: Which day has registered the lowest brent prince, on average, in last 19 years?

```python
temp = brent
temp['Dayname'] = temp.index.day_name()
temp.head()
```

```python
temp.loc['2011':]
```

```python
temp.Dayname.unique()
```

```python
temp.groupby('Dayname').mean().sort_values(by='Price')
```
