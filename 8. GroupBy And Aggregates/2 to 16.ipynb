### 10. Skill Challenge

#### 1. Create a smaller df from games, by selecting publisher, gener, platform, and NA_sales. Assign it to publisher

```python
games.info()
```

```python
publisher = games.iloc[:,[4,3,1,5]]
publisher.head()
```

#### 2. From the publisher df, find the top 10 game publish in North America by total sales

```python
publisher\
    .groupby('Publisher')\
    .sum(numeric_only=True)\
    .sort_values(
        by=['NA_Sales'], 
        ascending=False
    )\
    .head(10)
```

#### 3. Similarly, what the gaming platform that has attracted the most sales in North America?

```python
publisher\
    .groupby('Platform')\
    .sum(numeric_only=True)\
    .sort_values(
        by=['NA_Sales'], 
        ascending=False
    )
```

### 15. Fine-tuned Aggregates

```python
games.info()
```

```python
studios = games.iloc[:,[4,3,9]]
studios.head()
```

```python
studios.groupby(['Genre','Publisher']).sum()
```

```python
studios.groupby(['Genre','Publisher']).agg('sum')
```

```python
# Q: Summarize the sum and count of sales as well as the no. of games published by each publisher witin each genre
```

```python
studios.groupby(['Genre','Publisher'])\
    .agg(['sum','count'])\
    .sort_values(by=('Global_Sales','sum'),ascending=False)
```

```python
games[games.Publisher == 'Activision'].sort_values(by='Global_Sales', ascending=False).head(8)
```

### 16. Named Aggregations

```python
studios.groupby(['Genre','Publisher'])\
    .agg(['sum','count'])\
    .sort_values(by=('Global_Sales','sum'),ascending=False)\
    .rename({'sum':'total_sasles','count':'num_of_sales'},axis=1)
```

```python
# Name Aggregation
```

```python
studios.groupby(['Genre','Publisher'])\
    .agg(
        total_sales=('Global_Sales','sum'),
        num_of_sales=('Global_Sales','count')
    )
```

```python
# we can hand pick the desire columns
```

```python
games.groupby(['Genre','Publisher'])\
    .agg(
        total_global_sales=('Global_Sales','sum'),
        total_EU_sales=('EU_Sales','sum')
    )
```
