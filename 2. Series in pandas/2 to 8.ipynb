## 4. What's in The Data

```python
books_list = ['Fooled by Randumness', 'Sapiens', 'Lenin on the Train']
list_s = pd.Series(books_list)
print(list_s)
print(type(list_s))
```

```python
books_dict = {
    0:'Fooled by Randumness', 1:'Sapiens', 2:'Lenin on the Train'
}
dict_s = pd.Series(books_dict)
print(dict_s)
```

## 7. Index and RangeIndex

```python
books_list
```

```python
list_s
```

```python
pd.Series(data=books_list, index=["funny", "serious and amusing", "interesting"])
```

```python
list(pd.RangeIndex(start=1, stop=10, step=1))
```

```python
list(pd.RangeIndex(start=1, stop=-3, step=-1))
```

```python
range_index = list(pd.RangeIndex(start=1, stop=-2, step=-1))
pd.Series(data=books_list, index=range_index)
```
