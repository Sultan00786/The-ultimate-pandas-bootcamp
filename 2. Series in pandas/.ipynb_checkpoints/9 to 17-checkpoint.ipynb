### 9. Skill Challenge

```python
import pandas as pd
actor_name = ["Sharukh Khan", "Salman Khan", "Ritik", "Ranveer"]
actor_age = [55, 58, 53, 44]
print(actor_name)
print(actor_age)
```

```python
pd.Series(data = actor_age, index = actor_name)
```

```python
actor_dict = {}
for i in range(len(actor_name)):
    actor_dict[actor_name[i]] = actor_age[i]
actor_dict
```

```python
pd.Series(actor_dict)
```

### 13. Extracting By Index Position

```python
from string import ascii_uppercase as auc
auc
```

```python
alpha_series = pd.Series(data=list(auc), index=list(pd.RangeIndex(start =1,stop=len(auc)+1, step=1)))
alpha_series
```

#### 1. What is the first letter ?

```python
alpha_series[1]
```

#### 2. What is the 11th letter ?

```python
alpha_series[11]
```

#### 3. What are the first three letter ?

```python
alpha_series[:3]
```

#### 4. What are the six through tenth letter ?

```python
alpha_series[5:10]
```

#### 5. What are the last six letter ?

```python
alpha_series[len(alpha_series)-6:]
```

```python
alpha_series[-6:]
```

### 14. Accessing Elements By Label

```python
from string import ascii_lowercase as alc
alc
```

```python
lowercase_s = pd.Series(data=list(alc),index=list(auc))
lowercase_s
```

```python
lowercase_s = pd.Series(data=list(alc),index= ["label_" + i for i in auc])
lowercase_s
```

#### 1. What is the first letter ?

```python
lowercase_s['label_A']
```

#### 2. What is the 11th letter ?

```python
lowercase_s['label_K']
```

#### 3. What are the first three letter ?

```python
lowercase_s[:'label_C']
```

#### 4. What are the six through tenth letter ?

```python
lowercase_s['label_F':'label_J']
```

#### 5. What are the last six letter ?

```python
lowercase_s['label_U':]
```

### 17. Boolean Masks And The .loc Indexer

```python
lowercase_s[['label_U','label_U']]
```

```python
lowercase_s.loc[['label_U','label_U']]
```

```python
lowercase_s.loc[['label_U','label_U']]
```

```python
lower_bool_s = [True for i in lowercase_s]
print(lower_bool_s)
print("Size: ",len(lower_bool_s))
```

```python
lowercase_s[lower_bool_s]
```

```python
lower_bool_s = [True if (i%2==0) else False for i in range(len(lowercase_s))]
print(lower_bool_s)
print("Size: ",len(lower_bool_s))
```

```python
lowercase_s[lower_bool_s]
```

```python
lowercase_s[lower_bool_s].size
```
